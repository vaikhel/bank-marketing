# Exploratory Data Analysis

Analysis of the Bank Marketing dataset.  
The goal is to understand the data structure before modeling.

```python
import sys
sys.path.append("..")

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.preprocessing import load_data

sns.set_style("whitegrid")
```

```python
df = load_data()
df.head()
```

```python
print(f"Dataset shape: {df.shape}")
print("\nData types:")
print(df.dtypes)
print("\nNumber of missing values:", df.isnull().sum().sum())
```

```python
df.describe()
```

## Target Variable

```python
print(df["y"].value_counts())
print("\nIn percentages:")
print(df["y"].value_counts(normalize=True) * 100)

sns.countplot(data=df, x="y")
plt.title("Distribution of the target variable")
plt.show()
```

## Visualizations

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x="age", bins=30)
plt.title("Age distribution")
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x="job", order=df["job"].value_counts().index)
plt.xticks(rotation=45)
plt.title("Distribution by job")
plt.show()
```

```python
sns.boxplot(data=df, x="y", y="balance")
plt.title("Balance by subscription")
plt.show()
```
